## Easy Exercise 2

Ask the user for a number. Depending on whether the number is even or odd, print out an appropriate message to the user. Hint: How does an even / odd number react differently when divided by 2?

### Extras

- If the number is a multiple of 4, print out a different message.
- Ask the user for two numbers: one number to check (call it `num`) and one number to divide by (call it `check`).
- If `check` divides evenly into `num`, tell that to the user. If not, print a different appropriate message.

```python
num = int(input("Enter the number so I will check for you if even or odd"))
if num % 4 == 0:
    print(f"The number {num} is a multiple of 4.")
if num % 2 ==0:
    print(f"The number {num} is even")
else:
    print(f"The number {num} is odd")
```

```python
num = int(input("Enter a number"))
check = int(input("Enter a number you want to divided by"))

if num % check == 0:
    print(f"The number {num} is divisible by {check}")
else:
    print(f"The number {num} is not divisible by {check}")
```

## Easy Exercise 3

Write a program that takes a list of numbers (for example, `a = [5, 10, 15, 20, 25]`) and makes a new list of only the first and last elements of the given list.

For practice, write this code inside a function.

```python
def create_newlist(l):
    newlist=[l[0] , l[-1]]
    return newlist

a = [5, 10, 15, 20, 25]
print(create_newlist(a))
```

## Easy Exercise 4

Write a function that takes an ordered list of numbers (a list where the elements are in order from smallest to largest) and another number.

The function decides whether or not the given number is inside the list and returns (then prints) an appropriate boolean.

**Solution:** [Element Search](https://www.practicepython.org/exercise/2014/11/11/20-element-search.html)

```python
def ordered_list(l , n1):
    if n1 in l:
        return True
    else:
        return False 

numbers = [1 , 2 , 3 , 4 , 5]
num1 = 5

ordered_list(numbers , num1)
```
